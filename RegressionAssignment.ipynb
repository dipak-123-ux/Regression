1.	What is Simple Linear Regression?
    - 	Simple Linear Regression is a statistical method used to model and      quantify the relationship between two continuous variables: one independent variable (X) and one dependent variable (Y). The primary goal is to find a  straight line (the "best-fit" line) that minimizes the difference between  actual data points and the values predicted by the model.

2.	What are the key assumptions of Simple Linear Regression?
    	- For simple linear regression to produce valid predictions and reliable  statistical inferences, the data must satisfy four key assumptions. A  common and helpful acronym to remember these assumptions is LINE:  Linearity, Independence, Normality, and Equal Variance
      Linearity
      The relationship between the independent variable (𝑋) and the dependent variable (Y)
      must be linear. This means that a straight line is the most appropriate way to model
      Independence of Residuals
      The observations in the dataset must be independent of one another. This means that the error (residual) for one data point tells you nothing about the error for another data point.
      Normality of Residuals
      For any given value of X the errors (residuals) should be normally distributed
      Around
      the regression line. This means that small errors close to the line are very common, while
      massive errors are rare.
      Homoscedasticity (Equal Variance)
      The variance of the residuals must be constant across all levels of the independent
      variable. In simpler terms, the spread or "scatter" of the data points around the
      regression line should look uniform from the left side of the graph to the right side.

3.	What is heteroscedasticity, and why is it important to address in regression models?
    - 	Heteroscedasticity occurs when the variance of the residuals (the differences between observed and predicted values) is not constant across all levels of the independent variable.
While heteroscedasticity does not make the regression coefficients biassed (the line itself is still generally in the right place), it creates severe issues for the reliability of the model. Here is why it must be fixed:
Invalid Standard Errors
Standard regression algorithms (like Ordinary Least Squares) calculate the precision of their coefficients under the assumption that error variance is uniform. When heteroscedasticity is present, these calculations break down. The model will typically underestimate the standard errors for the predictors.
Flawed Hypothesis Testing(t-test and p-values)
Because standard errors are artificially low, the calculated t-statistics will be artificially
high. This leads to unreliable p-values causing you to conclude that an independent
variable is statistically significant when, in reality, it might just be random noise.
Inaccurate Confidence Intervals
Confidence and prediction intervals will be uniformly calculated using an average error variance. As a result, the intervals will be too wide in areas where the data is tightly clustered (making you look less precise than you are) and too narrow in areas where the data is highly scattered (giving a false sense of security).

4.	What is Multiple Linear Regression?
    -	Multiple Linear Regression (MLR) is a statistical technique used to model the relationship between one continuous dependent variable (Y) and two or more independent variables (X). While simple linear regression looks at how a single factor predicts an outcome, MLR allows you to isolate and examine the simultaneous impact of multiple factors at the same time.
The Core Equation
The mathematical model expands the simple linear regression formula to include multiple predictors:
\^y=β_0+β_1 x_1+β_2 x_2+⋯+β_n x_n+ε

ŷ (Dependent Variable): The target outcome you are trying to predict or explain.
x₁, x₂, ..., xₙ (Independent Variables): The distinct features or predictor variables used in the model.
β₀ (Intercept): The expected value of Y when all independent variables are equal to zero.
β₁, β₂, ..., βₙ (Partial Regression Coefficients): The unique effect of each predictor. For example, β₁ represents the expected change in Y for every one-unit increase in x₁, assuming all other independent variables stay completely constant.
ε (Error Term): The residual variance not explained by the model's predictors.

5.	What is polynomial regression, and how does it differ from linear regression?
    -	Polynomial regression is a form of regression analysis in which the relationship between the independent variable(x) and the dependent variable(y) is modeled as an nth degree polynomial.
While it is used to capture curved, non-linear relationships in data, it is technically considered a special case of multiple linear regression because the model is still linear in terms of its parameters (coefficients).
Key Differences: Linear vs. Polynomial Regression
Feature / Metric	Linear Regression	Polynomial Regression
Shape of the Fit	Strictly a straight line (or flat plane)	A curved line (can bend multiple times based on the degree)
Data Complexity	Best for simple, proportional trends	Best for complex, non-linear patterns (e.g., growth rates, physics trajectories)
Risk of Overfitting	Very low; can only underfit if data is curved	High; choosing too high a degree causes the line to wiggle wildly to hit every point
Extrapolation Risk	Moderately risky; assumes the stable trend continues indefinitely	Extremely dangerous; curves can shoot up or down toward infinity outside your data range

```python
#Q6. Implement a Python program to fit a Simple Linear Regression model to
#the following sample data:
#● X = [1, 2, 3, 4, 5]
#● Y = [2.1, 4.3, 6.1, 7.9, 10.2]
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Define the sample dataset
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Reshaped to 2D array as required by scikit-learn
Y = np.array([2.1, 4.3, 6.1, 7.9, 10.2])

# 2. Initialize and fit the Simple Linear Regression model
model = LinearRegression()
model.fit(X, Y)

# 3. Extract the calculated slope (beta_1) and intercept (beta_0)
slope = model.coef_[0]
intercept = model.intercept_

# 4. Generate predictions using the fitted model
predictions = model.predict(X)

# 5. Display the results
print("--- Regression Model Results ---")
print(f"Calculated Intercept (β₀): {intercept:.2f}")
print(f"Calculated Slope (β₁):     {slope:.2f}")
print(f"Final Model Equation:      Y = {intercept:.2f} + {slope:.2f} * X\n")

print("Actual Y vs Predicted Y:")
for act, pred in zip(Y, predictions):
    print(f"Actual: {act:<5} -> Predicted: {pred:.2f}")
```

```python
#Q7. Fit a Multiple Linear Regression model on this sample data:
#● Area = [1200, 1500, 1800, 2000]
#● Rooms = [2, 3, 3, 4]
#● Price = [250000, 300000, 320000, 370000]
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant

# 1. Prepare the dataset
df = pd.DataFrame({
    'Area': [1200, 1500, 1800, 2000],
    'Rooms': [2, 3, 3, 4],
    'Price': [250000, 300000, 320000, 370000]
})

X = df[['Area', 'Rooms']]
Y = df['Price']

# 2. Fit the Multiple Linear Regression model
model = LinearRegression()
model.fit(X, Y)

# 3. Calculate VIF for independent variables
# statsmodels requires an explicit intercept (constant column) to calculate standard VIF
X_with_constant = add_constant(X)

vif_data = pd.DataFrame()
vif_data["Feature"] = X.columns
vif_data["VIF"] = [
    variance_inflation_factor(X_with_constant.values, i)
    for i in range(1, X_with_constant.shape[1])
]

# 4. Display Results
print("--- Multiple Linear Regression Model ---")
print(f"Intercept (β₀):          {model.intercept_:.2f}")
print(f"Area Coefficient (β₁):   {model.coef_[0]:.2f}")
print(f"Rooms Coefficient (β₂):  {model.coef_[1]:.2f}\n")
print(f"Equation: Price = {model.intercept_:.2f} + ({model.coef_[0]:.2f} * Area) + ({model.coef_[1]:.2f} * Rooms)\n")

print("--- Multicollinearity Assessment (VIF) ---")
print(vif_data.to_string(index=False))

```

```python
#Q8. Implement polynomial regression on the following data:
#● X = [1, 2, 3, 4, 5]
#3
#● Y = [2.2, 4.8, 7.5, 11.2, 14.7]
#Fit a 2nd-degree polynomial and plot the resulting curve.
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# 1. Define the sample dataset
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([2.2, 4.8, 7.5, 11.2, 14.7])

# 2. Transform the features into 2nd-degree polynomial features (X and X^2)
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# 3. Initialize and fit the Linear Regression model on polynomial data
model = LinearRegression()
model.fit(X_poly, Y)

# 4. Extract parameters
intercept = model.intercept_
coef = model.coef_  # Format: [0, coefficient_for_X, coefficient_for_X^2]

print("--- Polynomial Regression Results ---")
print(f"Intercept (β₀):      {intercept:.2f}")
print(f"X Coefficient (β₁):   {coef[1]:.2f}")
print(f"X² Coefficient (β₂):  {coef[2]:.2f}")
print(f"Model Equation:       Y = {intercept:.2f} + {coef[1]:.2f}*X + {coef[2]:.2f}*X²\n")

# 5. Generate smooth curve lines for plotting
X_smooth = np.linspace(1, 5, 100).reshape(-1, 1)
X_smooth_poly = poly.transform(X_smooth)
Y_smooth_pred = model.predict(X_smooth_poly)

# 6. Plot the original data and the fitted polynomial curve
plt.figure(figsize=(8, 5))
plt.scatter(X, Y, color='red', label='Actual Data Points', zorder=5)
plt.plot(X_smooth, Y_smooth_pred, color='blue', linewidth=2, label='2nd-Degree Polynomial Fit')

plt.title('2nd-Degree Polynomial Regression')
plt.xlabel('Independent Variable (X)')
plt.ylabel('Dependent Variable (Y)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
#Q9. Create a residuals plot for a regression model trained on this data:
#● X = [10, 20, 30, 40, 50]
#● Y = [15, 35, 40, 50, 65]
#Assess heteroscedasticity by examining the spread of residuals
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Define the dataset
X = np.array([10, 20, 30, 40, 50]).reshape(-1, 1)
Y = np.array([15, 35, 40, 50, 65])

# 2. Fit the linear regression model
model = LinearRegression()
model.fit(X, Y)

# 3. Calculate predictions and residuals
predictions = model.predict(X)
residuals = Y - predictions

# 4. Print calculated values for verification
print("--- Model Statistics ---")
print(f"Equation: Y = {model.intercept_:.2f} + {model.coef_[0]:.2f} * X\n")
print("Predicted Y | Residuals:")
for pred, res in zip(predictions, residuals):
    print(f"Pred: {pred:<5.1f} | Residual: {res:+.1f}")

# 5. Generate the Residuals Plot
plt.figure(figsize=(8, 5))
plt.scatter(predictions, residuals, color='purple', s=100, label='Residuals')
plt.axhline(y=0, color='black', linestyle='--', linewidth=1.5, label='Zero Error Line')

plt.title('Residuals Plot (Predicted Values vs. Residuals)')
plt.xlabel('Predicted Values ($\hat{Y}$)')
plt.ylabel('Residuals ($Y - \hat{Y}$)')
plt.ylim(-7, 7)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.show()
```
